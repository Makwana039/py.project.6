```python
import os
from datetime import datetime

class journalManager:

    def __init__(self, filename="journal.txt"):
        self.filename = filename

    # 1. Add a New Entry
    def add_entry(self):
        try:
            entry = input("\nEnter your journal entry:")

            if entry.strip() == "":
                print("\nEntry cannot be empty!")
                return

            timestamp = datetime.now().strftime("%Y-%m-%d-%H:%M:%S")

            with open(self.filename,"a", encoding="utf-8") as file:
                file.write(f"[{timestamp}]\n")
                file.write(entry + "\n")
                file.write("-" * 40 + "\n")

            print("\nEntry added succesfully.")

        except PermissionError:
            print("\nError: permission denied.")
        except OSError as e:
            print("file error:", e)

    # 2. View All Entries
    def view_entries(self):
        try:
            with open(self.filename, "r") as file:
                data = file.read()

            if data.strip() == "":
                print("\nNo journal Entries found.")
            else:
                print("\n-----All Journal Entries-----")
                print(data)

        except FileNotFoundError:
            print("\nError: The Journal file does not exist.")
            print("Please add a new entrie file.")
        except PermissionError:
            print("\nError: permission denied.")
        except OSError as e:
            print("\nFile error:", e)

    # 3. Search for an entry
    def search_entry(self):
        try:
            keyword = input("Enter keyword to search: ")

            if keyword.strip() == "":
                print("Keyword cannot be empty!")
                return

            with open(self.filename, "r") as file:
                lines = file.readlines()

            found = False

            print("\n----- SEARCH RESULTS -----")

            for line in lines:
                if keyword.lower() in line.lower():
                    print(line.strip())
                    found = True

            if not found:
                print("No matching entry found.")

        except FileNotFoundError:
            print("No journal file exists yet.")
        except PermissionError:
            print("Error: Permission denied.")
        except Exception as e:
            print("Error:", e)

     # 4. Delete all entries
    def delete_all_entries(self):
        try:
            if not os.path.exists(self.filename):
                print("No journal file exists.")
                return

            confirm = input(
                "Are you sure you want to delete all entries? (yes/no): "
            )

            if confirm.lower() == "yes":

                # 'w' mode clears the file
                with open(self.filename, "w") as file:
                    file.write("")

                print("All journal entries deleted successfully!")

            else:
                print("Delete operation cancelled.")

        except PermissionError:
            print("Error: Permission denied.")
        except Exception as e:
            print("Error:", e)

 # Main menu
    def menu(self):

        while True:
            print("Welcome to personal Journal manager!")
            print("\n========== JOURNAL MANAGER ==========")
            print("1. Add a New Entry")
            print("2. View All Entries")
            print("3. Search for an Entry")
            print("4. Delete All Entries")
            print("5. Exit")
            print("=====================================")

            choice = input("Enter your choice: ")

            if choice == "1":
                self.add_entry()

            elif choice == "2":
                self.view_entries()

            elif choice == "3":
                self.search_entry()

            elif choice == "4":
                self.delete_all_entries()

            elif choice == "5":
                print("Thank you for using Journal Manager!")
                break

            else:
                print("Invalid choice! Please enter 1-5.")


# Object creation
journal = journalManager()

# Start program
journal.menu()
```
